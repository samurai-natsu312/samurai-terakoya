```python
# 人間の設計図（クラス）を定義する
class Human:

    # 名前と年齢の属性を初期化する
    def __init__(self, name, age):

        # 名前と年齢の属性を定義する
        self.name = name
        self.age = age

    # 名前と年齢を画面に出力するメソッドを定義する
    def printinfo(self):
        print(self.name)
        print(self.age)


# 設計図をもとに「A」と「1」を持つインスタンスを作成する
human = Human("A", 1)

# printinfoメソッドを実行して情報を表示する
human.printinfo()
```
