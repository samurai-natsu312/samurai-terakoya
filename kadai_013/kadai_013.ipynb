```python
# 与えられた引数priceに消費税を加算した計算結果を返す関数を定義する
def calculate_total(price, tax):
    # 与えられた引数priceに消費税を加算し、変数totalに代入する
    total = price + (price * (tax / 100))
    # 戻り値として変数totalの値を返す
    return total

# 第1引数に「110」、第2引数に「10」を設定して関数を呼び出し、戻り値を変数に代入する
result = calculate_total(110, 10)

# 変数resultの値を出力する
print(f"{result}円")
```
