```python
# 人間を表すHumanクラスを定義する
class Human:
    # 初期化メソッドで名前と年齢の属性を定義する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 年齢（属性）をもとに成人かどうかを判定して標準出力するメソッド
    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}は大人です")
        else:
            print(f"{self.name}は大人ではありません")


# 複数のインスタンスをまとめて管理する配列（リスト）を作成する
humans = [
    Human("A", 25),
    Human("B", 18),
    Human("C", 20)
]

# for文を使って、配列（リスト）の要素数分だけ順番に繰り返し処理を行う
for human in humans:
    # インスタンス化されたオブジェクトからメソッドを呼び出す
    human.check_adult()
```
